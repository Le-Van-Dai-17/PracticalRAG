### Embedding model

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name = "sentence-transformers/all-MiniLM-L6-v2"
)

embeddings
```

```python
text = "I am learning about RAG"
```

```python
converted = embeddings.embed_query(text)
converted
```

```python
list = [
    "Today is Tuesday",
    "I am learning about RAG",
    "I love Python programming language",
    "Python is my favorite language programming"
]
```

```python
list_Convert = embeddings.embed_documents(list)
list_Convert[1]
```

```python
converted
```

```python
import numpy as np
def cosine(vec1, vec2):
    dotProduct = np.dot(vec1, vec2)
    norm1 = np.linalg.norm(vec1)
    norm2 = np.linalg.norm(vec2)
    return dotProduct/(norm1 * norm2)
```

```python
print(list[0])
print(list[2])
print(cosine(list_Convert[0], list_Convert[1]))
```
