### PDF Loader

```python
from langchain_community.document_loaders import(
     PyPDFLoader, 
     PyMuPDFLoader
)
```

```python
pdf_loader = PyPDFLoader(
    "pdf/attention.pdf"
)

pdfDocument = pdf_loader.load()
print(pdfDocument[0].metadata)
print("==================================")
print(pdfDocument[0].page_content)
```

```python
print(len(pdfDocument))
```

```python
pdf_loader = PyMuPDFLoader(
    "pdf/attention.pdf"
)

pdfDocument = pdf_loader.load()
print(pdfDocument[0].metadata)
print("==================================")
print(pdfDocument[0].page_content)
```
