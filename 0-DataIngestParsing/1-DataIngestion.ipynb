### Introduction to Data Ingestion

```python
from langchain_core.documents import Document

doc = Document(
    page_content = "This is a simple document",
    metadata = {
        "source": "data/text_files/python_intro.txt",
        "author": "anyone",
        "topic": "any"
    }
)

print("Document")
print(doc.page_content)
print(doc.metadata["author"])
```

### Directory

```python
from langchain_community.document_loaders.directory import DirectoryLoader
```

```python
directoryLoader = DirectoryLoader(
    "data/text_files",
    "*.txt",
    loader_cls = TextLoader,
    show_progress = True
)

documents = directoryLoader.load()
```

```python
documents
```

### Text Splitting Text

```python
from langchain_text_splitters import (
    RecursiveCharacterTextSplitter,
    TokenTextSplitter, 
)
```

```python
len(documents[0].page_content)
```

```python
from langchain_text_splitters import CharacterTextSplitter
```

```python
character_splitter = CharacterTextSplitter(
    separator = "\n\n",
    chunk_size = 50,
    chunk_overlap = 20
)

character_splitter
```

```python
chunks = character_splitter.split_documents(documents)
chunks
```

```python
character_splitter2 = RecursiveCharacterTextSplitter(
    separators = ["\n\n", "\n", ".", ",", " "],
    chunk_size = 50,
    chunk_overlap = 20
)

chunks = character_splitter2.split_documents(documents)
chunks
```

```python
character_splitter3 = TokenTextSplitter(
    chunk_size = 50,
    chunk_overlap = 20
)

chunks = character_splitter3.split_documents(documents)

chunks

print(chunks[0].page_content)
print("+=====")
print(chunks[1].page_content)
```
